```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Escribe los tiempos uno por uno
t1 = 50
t2 = 95
t3 = 60
t4 = 120
t5 = 45

# Crea tu diccionario logístico
datos = {
    'tiempo_real': [t1, t2, t3, t4, t5],
    'cliente': ['Walmart', 'Soriana', 'Oxxo', 'Walmart', 'Oxxo'],
    'clima': ['Despejado', 'Lluvia', 'Despejado', 'Niebla', 'Lluvia']
}

df = pd.DataFrame(datos)

# Configura y dibuja el gráfico
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 4))
sns.barplot(x='cliente', y='tiempo_real', hue='clima', data=df, palette='magma')

# Agrega la línea roja de alerta en los 60 minutos
plt.axhline(y=60, color='red', linestyle='--')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Contamos cuantos envios hay por cada tipo de Status (Tu variable "Y")
print("--- RESUMEN DE STATUS DE ENVIOS ---")
print(df_real['Status'].value_counts())

# 2. creamos una grafica para ver el desempeño por Transportista (Carrier)
plt.figure(figsize=(10,5))
sns.countplot(data=df_real, x='Carrier', hue='Status', palette='viridis')

# 3. Personalizamos el reporte visual
plt.title('Desempeño de Entregas por Transportista (2,000 Envios)', fontsize=14, fontweigth='bold')
plt.xlabel('Transportista (Carrier)')
plt.ylabel('Cantidades de Envios')
plt.xticks(rotation=45) # Inclina los nombres para que no se encimen
plt.show()
```
